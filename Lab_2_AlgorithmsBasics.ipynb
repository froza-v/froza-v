The `func2(a, b)` function calculates `a` raised to the power of `b` (a^b) using repeated multiplication. It initializes `result` to 1 and then multiplies `result` by `a`, `b` times in a `while` loop.

### Time Complexity Analysis:

The `while` loop runs `b` times. Inside the loop, operations like multiplication, subtraction, and assignment take constant time, O(1). Therefore, the total time complexity is **O(b)**, which is linear with respect to the value of `b`.

```python
# Exercise 4
def func4(a, b):
    b.sort() # Time complexity for this function is O(Nlog N) where N = len(b)
    intersect = 0
    for x in a:
        if binary_search(b, x) >= 0:
            intersect = intersect + 1
    return intersect

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
# Sample function call: func4([1, 2, 3, 4], [2, 4, 6, 8])
print(func4([1, 2, 3, 4], [2, 4, 6, 8]))
```

The `func3(num)` function calculates the sum of the digits of a given number `num`. It initializes `result` to 0 and then iteratively adds the last digit of `num` to `result` and removes the last digit from `num` until `num` becomes 0.

### Time Complexity Analysis:

The `while` loop runs as many times as there are digits in `num`. The number of digits in `num` is proportional to `log10(num)`. Inside the loop, operations like modulo, addition, integer division, and assignment take constant time, O(1). Therefore, the total time complexity is **O(log(num))**, which is linear with respect to the number of digits in `num`.

```python
# Exercise 5
def func5(n):
    guess = 1
    while guess * guess <= n:
        if guess * guess == n:
            return guess
        guess = guess + 1
    return None
```

```python
# Sample function call: func5(16)
print(func5(16))
```

```python
# Exercise 6
def func8(A):
    func8_helper(A, 0, len(A) - 1)
    return A # Return the modified array

def func8_helper(A, i, j):
    # Input: An array A and two indices i and j.
    n = j - i + 1 # Correct calculation of the length of the sub-array A[i...j]

    if n < 2: # Base case: if length is 0 or 1, it's already sorted
        return
    elif n == 2:
        if A[i] > A[j]:
            # Swap A[i] and A[j]
            A[i], A[j] = A[j], A[i]
    else: # n > 2, recursive step
        m = n // 3
        # Sort the first 2/3 of the array
        func8_helper(A, i, j - m)
        # Sort the last 2/3 of the array
        func8_helper(A, i + m, j)
        # Sort the first 2/3 again to ensure elements are in place
        func8_helper(A, i, j - m)
    # The helper function modifies A in place, so no return value is needed from it.
```

```python
#Sample function call: func8([3,1,2,7,8,6])
print(func8([3,1,2,7,8,6]))
```

The `func4(a, b)` function calculates the number of common elements (intersection) between two arrays `a` and `b`. It first sorts array `b` (using `b.sort()`) and then iterates through each element `x` in array `a`. For each `x`, it performs a binary search in the sorted array `b` to check for its presence. If `x` is found in `b`, it increments a counter `intersect`.

### Time Complexity Analysis:

1.  **Sorting `b`**: If `N` is the length of array `b`, sorting takes **O(N log N)** time.
2.  **Iterating through `a`**: If `M` is the length of array `a`, the loop runs `M` times.
3.  **Binary search**: Inside the loop, `binary_search(b, x)` takes **O(log N)** time because `b` is sorted.

Therefore, the total time complexity is **O(N log N + M log N)**. In the worst case, if `N` and `M` are similar in size, this simplifies to **O(N log N)** or **O(M log M)**, where N and M are the lengths of the respective arrays.

The `func5(n)` function attempts to find the integer square root of `n`. It starts with `guess = 1` and iteratively increments `guess` until `guess * guess` is greater than `n`. If at any point `guess * guess` equals `n`, it returns `guess`. If the loop finishes without finding an exact square root, it returns `None`.

### Time Complexity Analysis:

The `while` loop continues as long as `guess * guess <= n`. In the worst case, `guess` will increment up to `sqrt(n)`. Inside the loop, operations like multiplication, comparison, and addition take constant time, O(1). Therefore, the total time complexity is **O(sqrt(n))**, which is linear with respect to the square root of `n`.

```python
# Exercise 7
def func12(arr, k):
    arr_len = len(arr)
    k = k % arr_len
    # Initialize a new array of the same length as 'arr'
    custom_arr = [0] * arr_len

    for i in range(arr_len):
        mystery_index = (i + k) % arr_len
        custom_arr[mystery_index] = arr[i]
    return custom_arr

```

```python
# write a function named as Binary_Search to perform binary search in 1D-array
def Binary_Search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid  # Return the index if target is found
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1 # Return -1 if target is not found
```

```python
# call the function Binary_Search for array arr = [1, 3, 5, 7, 9,15,16,14,45]
# Note: Binary Search requires a sorted array. The provided sample array is not fully sorted.
# Let's use a sorted array for demonstration.
arr = [1, 3, 5, 7, 9, 14, 15, 16, 45]
target_found = 15
index = Binary_Search(arr, target_found)
if index != -1:
    print(f"Element {target_found} found at index {index}")
else:
    print(f"Element {target_found} not found in the array")

target_not_found = 10
index = Binary_Search(arr, target_not_found)
if index != -1:
    print(f"Element {target_not_found} found at index {index}")
else:
    print(f"Element {target_not_found} not found in the array")
```
